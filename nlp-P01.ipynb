## Exploratory Data Analysis

In the following 5 sections, the objective will be to replicate a series of visual plots using the "sou" dataframe and various python tools.

### Number of Speeches per President

```python
speech_count = sou["President"].value_counts()
data_merged = pd.merge(sou, speech_count, how="inner", on="President")
final_frame = data_merged[["President", "count"]].groupby("President", sort=False).agg("first").reset_index()
```

```python
plt.figure(figsize=(14, 6))
plt.bar(final_frame["President"], final_frame["count"], width=0.5)
plt.title("Number of Speeches per President")
plt.xlabel("President")
plt.ylabel("count")
plt.xticks(rotation=90)
plt.show()
plt.savefig("outputs/number_speeches_president.png")
```

In this plot, the number of speeches per president in US History is displayed. There seems to be a more or less uniform distribution of speech counts across all Presidents. George Washington at the hgihest at 12 makes sense given the necessity of speeches when forming a country for the first time. Similarly with FDR, we see a spike in speeches as needed to address various crisis such as the Great Depression & WW2 to the public.
