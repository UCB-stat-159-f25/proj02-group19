# Part 1: Data Loading and Initial Exploration

## Data

In this section, we are reading in the sotu csv file located in the data directory as a pandas dataframe to perform data analysis. The table contains data on the speech and their word count for various presidents since the foudning of the United States.

```python
# imports

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-dark') 
```

```python
sou = pd.read_csv("data/SOTU.csv")
```

```python
sou
```

## Exploratory Data Analysis

In the following 5 sections, the objective will be to replicate a series of visual plots using the "sou" dataframe and various python tools.

### Number of Speeches per President

```python
speech_count = sou["President"].value_counts()
data_merged = pd.merge(sou, speech_count, how="inner", on="President")
final_frame = data_merged[["President", "count"]].groupby("President", sort=False).agg("first").reset_index()
```

```python
plt.figure(figsize=(14, 6))
plt.bar(final_frame["President"], final_frame["count"], width=0.5)
plt.title("Number of Speeches per President")
plt.xlabel("President")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.savefig("outputs/number_speeches_president.png")
```

In this plot, the number of speeches per president in US History is displayed. There seems to be a more or less uniform distribution of speech counts across all Presidents. George Washington at the hgihest at 12 makes sense given the necessity of speeches when forming a country for the first time. Similarly with FDR, we see a spike in speeches as needed to address various crisis such as the Great Depression & WW2 to the public.

### Number of Speeches per Year

```python
grouped_frame = sou[["Year", "Text"]].groupby("Year").agg("count").reset_index()
grouped_frame = grouped_frame.rename(columns={"Text": "Count"})
```

```python
plt.figure(figsize=(8, 6))
plt.plot(grouped_frame["Year"], grouped_frame["Count"])
plt.title("Number of State of the Union Speeches per Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.savefig("outputs/number_speeches_per_year.png")
```

In this plot, we see the total number of speeches given per year. There seems to be a masisve spike during the first year of the United States, given the nature of the situation and the uncertainities surrounding the founding year. Most years tend to have one speech, which is typical for the State of the Union speech. We do see spikes durng the late 1900s, given the trubulent times such as several wars, crisis, economic downturns that could explain why more than one speech was needed during those years.

### Word Count Distribution

```python
plt.figure(figsize=(8, 6))
sns.histplot(sou, x="Word Count")
plt.title("Distribution of the State of the Union Speech Word Counts")
plt.xlabel("Word Count")
plt.ylabel("Count")
plt.savefig("outputs/word_distribution.png")
```

This plot displays the general shape of the how long (in terms of word count) different state of the union speeches are. Its clear that the graph is skewed right with a peak aorunf the 5000 word mark. Very few state of the union speeches in US history cross 25,000 words.

### Word Count Distribution over Year

```python
plt.figure(figsize=(8, 6))
sns.rugplot(sou, x="Word Count", y="Year")
sns.scatterplot(sou, x="Word Count", y="Year")
plt.title("Speech Year versus Word Count")
plt.xlabel("Word Count")
plt.ylabel("Year")
plt.savefig("outputs/word_distribution_over_year.png")
```

This visual displays an overlaid plot of a rugplot along with a scatterplot. This helps us identify any relationships/patterns that exist between the the year the speech was given and the word count in those speeches. There seems to be a slight positive correlation between the variables – as word count with longer speeches becoming more and more common as we approach the modern era. The rugplot adds another dimension of understanding as the distribution of the points in specfic regions/clusters aids in our scatterplot analysis.

### Word Count Distribution per President

```python
sou_grouped = sou[["President", "Word Count"]].groupby("President", sort=False).agg("mean").reset_index()
```

```python
plt.figure(figsize=(14, 6))
plt.bar(sou_grouped["President"], sou_grouped["Word Count"], width=0.5)
plt.title("Average State of the Union Word Count per President")
plt.xlabel("President")
plt.ylabel("Average Word Count")
plt.xticks(rotation=90)
plt.savefig("outputs/avg_speech_word_count.png")
```

The general trend of average word count shows minimal values during the foundation of the country and gradually increases to a peak around Teddy Roosevelt's presidency. The average counts begin falling again as we approach the modern era. Most speeches will be of similar average word count since these are timed speeches and the content format/style will not vary too much. Perhaps the peaks are reflective of special occasions or periods of uncertainty that required more attention (ex. World War 1).
