```python
# import libraries

import spacy
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Make plots consistent
plt.style.use('seaborn-v0_8-dark')

# Load spaCy model with word vectors
# Note: en_core_web_sm has no vectors → so similarity is worse.
# If allowed, use en_core_web_md. If not, keep sm.
nlp = spacy.load("en_core_web_md")
```

```python
# clean data and prepare it
sou = pd.read_csv("data/SOTU.csv")

sou_df = sou.copy()
sou_df = sou_df[sou_df["Year"] >= 2000] # modern speeches
```

```python
# Combine speeches by president
pres_speeches = sou_df.groupby("President")["Text"].apply(lambda texts: " ".join(texts)).to_dict()
```

```python
# Process combined speeches into spaCy docs
pres_docs = {pres: nlp(text) for pres, text in pres_speeches.items()}
```

```python
presidents = list(pres_docs.keys())
pairs = []

# computing the similarity of each pair of presidents
for i in range(len(presidents)):
    for j in range(i + 1, len(presidents)):
        p1, p2 = presidents[i], presidents[j]
        sim = pres_docs[p1].similarity(pres_docs[p2])
        pairs.append((p1, p2, sim))

# sorting pairs by similarity in descending order
pairs_sorted = sorted(pairs, key=lambda x: x[2], reverse=True)

most_similar = pairs_sorted[0]

print("Most similar presidents (2000+ speeches):")
print(f"{most_similar[0]} and {most_similar[1]} → similarity score = {most_similar[2]:.3f}")
```
